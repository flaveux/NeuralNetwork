```python
import numdifftools as nd
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(-1, 1.1, 0.01)
def f(x):
    return x ** 9 + 15 * x ** 5 - 14 * x ** 2 + 9

plt.plot(x, f(x))
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()

arr_of_proizv = np.arange(0.1, 1.1, 0.01) # x0 = 0.1 отсюда старт (нужно было для массив производных)

proizv = nd.Derivative(f) # nd.Gradient возвращал 99 массивов с одним значением
# proizv само применение производной не возвращает массив
# arr_proizv = proizv(arr_of_proizv)
# print(arr_proizv) # значения производой в точках
# ???

h = 0.01
a_start = 1 # опираясь на график, с точки -1 после 3 итерации выдал значние а в триллион (свалился влево)

for i in range(30): # 30 итераций по первому графику эмперически
    dot_proivz = proizv(a_start) # значение производной в точке а_старт
    change = -h * dot_proivz
    a_next = a_start + change
    print(f"Новое значение a: {a_next:.5f}, изменения: {change:.6f}")
    a_start = a_next

# у этой функции нет глобального минимума, был найден локальный за 15 итераций
```
