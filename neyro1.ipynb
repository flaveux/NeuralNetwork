```python
!pip install numdifftools
import pandas as pd
import numpy as np
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics
import matplotlib as plt
import numdifftools as nd
df = pd.read_csv('https://raw.githubusercontent.com/rvgarafutdinov/python_book_2023/main/lab7_datasets/dataset_12.csv')

X_cols = df[['Class', 'Year_begin']]
Y_col = df['TeamSize']

X_train, X_test, y_train, y_test = train_test_split(X_cols, Y_col, test_size=0.2, random_state=12)
model1 = LinearRegression()
model1.fit(X=X_train, y=y_train) #fit - training
print(f"Коэффициенты модели: {model1.coef_}, свободный член: {model1.intercept_}")

y_pred_m1 = model1.predict(X_test)
r2_test = metrics.r2_score(y_test, y_pred_m1)

class CreateModel:
    def __init__(self, x = 0): #constructor
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X, y):
        X_b = np.c_[np.ones(X.shape[0]), X] # добавление единичного столбика
        coef_matrix = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
        self.intercept_ = coef_matrix[0]
        self.coef_ = coef_matrix[1:]
        print(coef_matrix)
    def predict(self, X):
        return self.intercept_ + np.dot(X, self.coef_)

model2 = CreateModel()
model2.fit(X_train, y_train)
y_pred_m2 = model2.predict(X_test)

r2_m2 = metrics.r2_score(y_test, y_pred_m2)
r2_m1 = metrics.r2_score(y_test, y_pred_m1)
print("R2:")
print(r2_m2, r2_m1)

print("КФ:")
print(f"model1 - Intercept: {model2.intercept_}, Coef: {model2.coef_}")
print(f"model2(sklearn) - Intercept: {model1.intercept_}, Coef: {model1.coef_}")

X = [(3.0, 2.1)
(4.2, 5.5)
(4.9, 9.2)]

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, cmap='viridis') # cmap - цветовая карта

ax.set_xlabel('Ось X')
ax.set_ylabel('Ось Y')
ax.set_zlabel('Ось Z')

plt.show()
```
